# IPEDS LEADERSHIP DATA PREPROCESSING

```python
import pandas as pd

df1 = pd.read_csv("S2023_IS.csv")
df1.head(10)
```

```python
# Filter rows where the 'SISCAT' column has specific values (prof, associate prof, tenured)
filter_values = [101, 102, 200, 201, 202, 203]
filtered_df = df1[df1['SISCAT'].isin(filter_values)]

# Sort the filtered data by the 'UNITID' column in ascending order
sorted_df = filtered_df.sort_values(by='UNITID', ascending=True)

# Display the result
sorted_df.head(10)
```

```python
sorted_df.to_csv('S2023_IS_sorted.csv', index=False)
```

# IPEDS DIVERSITY DATA PREPROCESSING

```python
df2 = pd.read_csv("S2023_OC.csv")
df2.head(20)
```

```python
# Filter rows where the 'OCCUPCAT' column has specific values (non-teaching staff)
filter_values = [220, 230, 250, 300, 310, 320, 330, 340, 350, 360, 370, 380, 390, 420, 490]
filtered_df = df2[df2['OCCUPCAT'].isin(filter_values)]

# Sort the filtered data by the 'UNITID' column in ascending order
sorted_df = filtered_df.sort_values(by='UNITID', ascending=True)

# Display the result
sorted_df.head(20)
```

```python
# Group by the 'UNITID' column and sum all numerical columns
combined_df = sorted_df.groupby('UNITID', as_index=False).sum()

# Drop irrelevant columns
combined_df = combined_df.drop(columns=['FTPT', 'SABDTYPE', 'STAFFCAT'])

# Set all values in the column 'OCCUPCAT' to 100
combined_df['OCCUPCAT'] = 100

# Display the resulting DataFrame
combined_df.head(20)
```

```python
# Save the result to a new CSV file
combined_df.to_csv('S2023_OC_sorted.csv', index=False)
```

# IPEDS POLICY DATA PREPROCESSING

```python
df3 = pd.read_csv("SAL2023_IS.csv")
df3.head(20)
```

```python
# Filter rows where the 'ARANK' column has specific values (non-teaching staff)
filter_values = [7]
filtered_df = df3[df3['ARANK'].isin(filter_values)]

# Sort the filtered data by the 'UNITID' column in ascending order
sorted_df = filtered_df.sort_values(by='UNITID', ascending=True)

# Display the result
sorted_df.head(10)
```

```python
# Keep only columns useful
sorted_df = sorted_df[['UNITID', 'ARANK', 'SAINSTM', 'SAINSTW', 'SAOUTLM', 'SAOUTLW']]

sorted_df.head()
```

```python
sorted_df['AVGSALM'] = sorted_df['SAOUTLM'] / sorted_df['SAINSTM']
sorted_df['AVGSALW'] = sorted_df['SAOUTLW'] / sorted_df['SAINSTW']

sorted_df.head()
```

```python
sorted_df['AVGSALM'] = sorted_df['AVGSALM'].round(2)
sorted_df['AVGSALW'] = sorted_df['AVGSALW'].round(2)

sorted_df.head()
```

```python
sorted_df.to_csv('SAL2023_IS_sorted.csv', index=False)
```

# IPEDS DAYCARE DATA PREPROCESSING

```python
df4 = pd.read_csv("IC2023.csv")
df4.head(10)
```

```python
# Filter rows where 'LEVEL' equals 1 and 'CNTLAFFI' is in [1, 3, 4]
filtered_df = df4[(df4['LEVEL3'] == 1) |
 (df4['LEVEL4'] == 1) | (df4['LEVEL5'] == 1) |
 (df4['LEVEL6'] == 1) | (df4['LEVEL7'] == 1) |
 (df4['LEVEL8'] == 1) | (df4['LEVEL17'] == 1) |
 (df4['LEVEL18'] == 1) | (df4['LEVEL19'] == 1)]

filtered_df = filtered_df[(filtered_df['CNTLAFFI'].isin([1, 3, 4]))]

# Display the filtered DataFrame
filtered_df.head(10)
```

```python
filtered_df = filtered_df[['UNITID', 'CNTLAFFI', 'LEVEL3', 'LEVEL4',
                           'LEVEL5', 'LEVEL6', 'LEVEL7', 'LEVEL8',
                           'LEVEL17', 'LEVEL18', 'LEVEL19', 'STUSRV8']]

filtered_df.head(10)
```

```python
filtered_df.to_csv('IC2023_sorted.csv', index=False)
```
