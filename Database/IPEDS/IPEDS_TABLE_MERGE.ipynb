# IPEDS TABLE MERGE

## Filtering schools based on our requirements

```python
import pandas as pd

table1 = pd.read_csv("C2023_B_final.csv")
table2 = pd.read_csv("EF2023_final.csv")
table3 = pd.read_csv("IC2023_sorted.csv")
```

```python
# Merge the first two tables on 'UNITID'
merged_df = pd.merge(table1, table2, on='UNITID', how='inner')

# Merge the result with the third table
filter_df = pd.merge(merged_df, table3, on='UNITID', how='inner')

filter_df.head(10)
```

```python
filter_df.shape
```

```python
filter_df = filter_df[filter_df['CSTOTLT'] >= 50]
filter_df.shape
```

```python
filter_df = filter_df[filter_df['EFTOTAL'] >= 200]
filter_df.shape
```

```python
filter_df.head(10)
```

```python
filter_df.to_csv('filter_final.csv', index=False)
```

## Merge with HD2023

```python
hd2023 = pd.read_csv("HD2023.csv")
hd2023.head(10)
```

```python
merged = pd.merge(filter_df, hd2023, on='UNITID', how='inner')
merged.head(10)
```

```python
merged.shape
```

## Continue with S2023_IS_final

```python
df = pd.read_csv("S2023_IS_final_decimal.csv")
df.head(10)
```

```python
df = df.drop(columns=['HRTOTLT', 'HRTOTLM', 'HRTOTLW'])
df.head()
```

```python
# Step 2: Reshape the data with pivot
# 'UNITID' becomes the index, 'SISCAT' values become column suffixes
pivoted_df = df.pivot(index="UNITID", columns="SISCAT", values=[col for col in df.columns if col not in ["UNITID", "SISCAT"]])
```

```python
# Step 3: Rename the columns
# Flatten the multi-level column index created by pivot
pivoted_df.columns = [f"{col[0]}_{int(col[1])}" for col in pivoted_df.columns]
```

```python
pivoted_df.reset_index(inplace=True)
```

```python
pivoted_df.fillna(0, inplace=True)
```

```python
pivoted_df.head(10)
```

```python
pivoted_df.to_csv('S2023_IS_FINAL.csv', index=False)
```
