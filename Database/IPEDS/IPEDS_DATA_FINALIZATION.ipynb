# IPEDS DATA FINALIZATION FOR EACH TABLE



## S2023_IS

```python
import pandas as pd

df1 = pd.read_csv("S2023_IS_sorted_decimal.csv")
df1.head(10)
```

```python
filter_values = [101, 102, 200]
filtered_df = df1[df1['SISCAT'].isin(filter_values)]
filtered_df.head(10)
```

```python
# Drop irrelevant columns
filtered_df = filtered_df[['UNITID', 'SISCAT', 'HRTOTLT', 'HRTOTLM',
                           'HRTOTLW', 'PRFPERCW', 'BKAAPERCW', 'HISPERCW',
                           'AISAPERCW']]
filtered_df.head(10)
```

```python
filtered_df.to_csv('S2023_IS_final_decimal.csv', index=False)
```
