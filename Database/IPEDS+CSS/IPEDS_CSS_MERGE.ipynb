# IPEDS CSS MERGE

## Merge CSS (hate & vawa)

```python
import pandas as pd

hate = pd.read_csv("hate_final_202122.csv")
vawa = pd.read_csv("vawa_final_202122.csv")
```

```python
css = pd.merge(hate, vawa, on='OPEID', suffixes=('', '_dup'))
css.head()
```

```python
css.to_csv("css_final_202122.csv", index=False)
```

## Merge IPEDS & CSS

```python
css = pd.read_csv("css_final_202122.csv")
ipeds = pd.read_csv("merged_final.csv")
```

```python
merged_df = pd.merge(ipeds, css, on='OPEID', how='inner')
merged_df.head(10)
```

```python
merged_df.shape
```

```python
merged_df.to_csv("merged_ipeds_css.csv", index=False)
```
