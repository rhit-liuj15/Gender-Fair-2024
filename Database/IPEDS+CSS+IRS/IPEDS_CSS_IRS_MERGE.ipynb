```python
import pandas as pd

ipeds_css = pd.read_csv("merged_ipeds_css_decimal.csv")
irs = pd.read_csv("irs990.csv")
ipeds_css.head()
```

```python
merged_df = pd.merge(ipeds_css, irs, on='EIN', how='left')
merged_df.head(10)
```

```python
merged_df.shape
```

```python
merged_df.rename(columns={'INSTNM_x': 'INSTNM'}, inplace=True)
merged_df = merged_df.drop("INSTNM_y", axis=1)
merged_df.head()
```

```python
merged_df.shape
```

Renaming Columns metadata

```python
merged_df.rename(columns={'CSTOTLT': 'DEGREETOT'}, inplace=True)
merged_df.rename(columns={'EFTOTAL': 'ENROLLTOT'}, inplace=True)
merged_df.rename(columns={'SAINSTM': 'ACADEMICPOPM'}, inplace=True)
merged_df.rename(columns={'SAINSTW': 'ACADEMICPOPF'}, inplace=True)
merged_df.rename(columns={'SAOUTLM': 'SALARYTOTM'}, inplace=True)
merged_df.rename(columns={'SAOUTLW': 'SALARYTOTF'}, inplace=True)
merged_df.rename(columns={'AVGSALM': 'SALARYPPM'}, inplace=True) #
merged_df.rename(columns={'AVGSALW': 'SALARYPPF'}, inplace=True) #
merged_df.rename(columns={'OC_BLKPERC': 'NONACD_BLKPCT'}, inplace=True)
merged_df.rename(columns={'OC_HISPPERC': 'NONACD_HSPPCT'}, inplace=True)
merged_df.rename(columns={'OC_ASIAPERC': 'NONACD_ASIAPCT'}, inplace=True)
merged_df.rename(columns={'PRFPERCW_101': 'PROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'PRFPERCW_102': 'ASSOCPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'PRFPERCW_200': 'TENUREWOMENPCT'}, inplace=True)
merged_df.rename(columns={'BKAAPERCW_101': 'BLACKPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'BKAAPERCW_102': 'BLACKASSOCPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'BKAAPERCW_200': 'BLACKTENUREWOMENPCT'}, inplace=True)
merged_df.rename(columns={'HISPERCW_101': 'HISPANICPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'HISPERCW_102': 'HISPANICASSOCPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'HISPERCW_200': 'HISPANICTENUREWOMENPCT'}, inplace=True)
merged_df.rename(columns={'AISAPERCW_101': 'ASIANPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'AISAPERCW_102': 'ASIANASSOCPROFWOMENPCT'}, inplace=True)
merged_df.rename(columns={'AISAPERCW_200': 'ASIANTENUREWOMENPCT'}, inplace=True)
merged_df.rename(columns={'AVGPERY_HATE': 'YEARLYHATECRIME'}, inplace=True)
merged_df.rename(columns={'PERTHOU_HATE': 'YEARLYHATECRIME1K'}, inplace=True)
merged_df.rename(columns={'AVGPERY_VAWA': 'YEARLYVAWA'}, inplace=True)
merged_df.rename(columns={'PERTHOU_VAWA': 'YEARLYVAWA1K'}, inplace=True)
merged_df.rename(columns={'STUSRV8': 'HASDAYCARE'}, inplace=True)
merged_df.head()
```
