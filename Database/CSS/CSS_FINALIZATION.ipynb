```python
import pandas as pd

df1 = pd.read_csv("hate_cleaned_202122.csv")
df2 = pd.read_csv("vawa_cleaned_202122.csv")
df1.head()
```

```python
df1.rename(columns={'AVG_CRIME': 'AVGPERY_HATE'}, inplace=True)
df2.rename(columns={'AVG_CRIME': 'AVGPERY_VAWA'}, inplace=True)
```

```python
df1['PERTHOU_HATE'] = (df1['AVGPERY_HATE'] / df1['Total']) * 1000
df2['PERTHOU_VAWA'] = (df2['AVGPERY_VAWA'] / df2['Total']) * 1000
df2.head()
```

```python
df2.shape
```

```python
df1['AVGALL_HATE'] = df1['AVGPERY_HATE'].sum() / 5684
df2['AVGALL_VAWA'] = df2['AVGPERY_VAWA'].sum() / 5682
df2.head()
```

```python
df1.to_csv('hate_final_202122.csv', index=False)
df2.to_csv('vawa_final_202122.csv', index=False)
```
