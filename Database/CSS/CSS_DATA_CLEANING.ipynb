#HATE/VAWA DATA LOADING

```python
import pandas as pd

df1 = pd.read_csv("Oncampushate202122.csv")
df1.head()
```

```python
df2 = pd.read_csv("Oncampusvawa202122.csv")
df2.head()
```

#Dropping irrelevant columns

```python
df1.drop(columns=['BRANCH', 'Address', 'ZIP', 'sector_cd', 'UNITID_P', 'FILTER20', 'FILTER21', 'FILTER22'], inplace=True)
df2.drop(columns=['BRANCH', 'Address', 'ZIP', 'sector_cd', 'UNITID_P', 'FILTER20', 'FILTER21', 'FILTER22'], inplace=True)
print(df1.shape)
print(df2.shape)
df1.head()
```

#Merging schools that have multiple entries into one

```python
# List of columns to apply 'first'
first_columns = ['OPEID', 'City', 'State', 'Sector_desc', 'men_total', 'women_total', 'Total']
numeric_cols1 = df1.select_dtypes(include=['number']).columns.tolist()
numeric_cols2 = df2.select_dtypes(include=['number']).columns.tolist()

# Generate the aggregation dictionary
agg_dict1 = {col: 'first' for col in first_columns}  # Apply 'first' for specific columns
agg_dict1.update({col: 'sum' for col in numeric_cols1 if col not in first_columns})  # Apply 'sum' for others

agg_dict2 = {col: 'first' for col in first_columns}  # Apply 'first' for specific columns
agg_dict2.update({col: 'sum' for col in numeric_cols2 if col not in first_columns})  # Apply 'sum' for others

hate = df1.groupby('INSTNM', as_index=False).agg(agg_dict1)
vawa = df2.groupby('INSTNM', as_index=False).agg(agg_dict2)
print(hate.shape)
print(vawa.shape)
```

# Combining crimes over the three years into one column, and compute their average

```python
hate['AVG_crimes'] = hate.iloc[:, 8:385].sum(axis=1)
vawa['AVG_crimes'] = vawa.iloc[:, 8:16].sum(axis=1)
vawa.head()
```

```python
vawa = vawa.drop(vawa.columns[8:17], axis=1)
hate = hate.drop(hate.columns[8:386], axis=1)
hate.head()
```

```python
hate['AVG_crimes'] = hate['AVG_crimes'] / 3
vawa['AVG_crimes'] = vawa['AVG_crimes'] / 3
vawa.head()
```

```python
hate.to_csv('hate_cleaned_202122.csv', index=False)
vawa.to_csv('vawa_cleaned_202122.csv', index=False)
```
